```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Create figure and axis
fig, ax = plt.subplots(1, 1, figsize=(14, 18))
ax.set_xlim(0, 10)
ax.set_ylim(0, 20)
ax.axis('off')

# Define colors
color_start = '#90EE90'  # Light green
color_process = '#87CEEB'  # Sky blue
color_decision = '#FFD700'  # Gold
color_end = '#FFB6C1'  # Light pink
color_activity = '#DDA0DD'  # Plum

# Helper function to create boxes
def create_box(ax, x, y, width, height, text, color, shape='round'):
    if shape == 'round':
        box = FancyBboxPatch((x - width/2, y - height/2), width, height,
                            boxstyle="round,pad=0.1", 
                            facecolor=color, edgecolor='black', linewidth=2)
    elif shape == 'diamond':
        # Diamond shape for decision
        points = [(x, y + height/2), (x + width/2, y), 
                 (x, y - height/2), (x - width/2, y)]
        box = patches.Polygon(points, facecolor=color, edgecolor='black', linewidth=2)
    else:
        box = FancyBboxPatch((x - width/2, y - height/2), width, height,
                            boxstyle="square,pad=0.1",
                            facecolor=color, edgecolor='black', linewidth=2)
    
    ax.add_patch(box)
    ax.text(x, y, text, ha='center', va='center', fontsize=10, 
            fontweight='bold', wrap=True)

# Helper function to create arrows
def create_arrow(ax, x1, y1, x2, y2, label=''):
    arrow = FancyArrowPatch((x1, y1), (x2, y2),
                           arrowstyle='->', mutation_scale=20, 
                           linewidth=2, color='black')
    ax.add_patch(arrow)
    if label:
        mid_x, mid_y = (x1 + x2) / 2, (y1 + y2) / 2
        ax.text(mid_x + 0.3, mid_y, label, fontsize=9, 
               bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray'))

# Title
ax.text(5, 19.5, 'KAARTEN GAME FLOWCHART', ha='center', va='center', 
        fontsize=16, fontweight='bold')

# 1. START
create_box(ax, 5, 18.5, 2, 0.6, 'START', color_start, 'round')

# Arrow down
create_arrow(ax, 5, 18.2, 5, 17.5)

# 2. MENU (Main Menu)
create_box(ax, 5, 17, 2.5, 0.8, 'MAIN MENU', color_process, 'round')

# Arrow down
create_arrow(ax, 5, 16.6, 5, 15.9)

# 3. Synopsis/Tutorial (New Player)
create_box(ax, 5, 15.5, 2.5, 0.6, 'Synopsis/Tutorial\n(New Player)', color_process, 'round')

# Arrow down
create_arrow(ax, 5, 15.2, 5, 14.5)

# 4. Menu Options (Decision Diamond)
create_box(ax, 5, 14, 2.5, 1, 'Menu Options:\nShop/Collection/\nMissions/Profile', color_decision, 'diamond')

# Arrows to different paths
create_arrow(ax, 3.8, 13.5, 2.5, 13, 'Shop')
create_arrow(ax, 6.2, 13.5, 7.5, 13, 'Collection')
create_arrow(ax, 5, 13.5, 5, 12.5, 'Missions')

# LEFT PATH - SHOP
create_box(ax, 2.5, 12.5, 1.8, 0.6, 'Open Shop', color_activity, 'round')
create_arrow(ax, 2.5, 12.2, 2.5, 11.5)
create_box(ax, 2.5, 11, 1.8, 0.8, 'Buy Pack\n(Spend Money)', color_activity, 'round')
create_arrow(ax, 2.5, 10.6, 2.5, 9.9)

# MIDDLE PATH - MISSIONS
create_box(ax, 5, 12, 1.8, 0.6, 'Daily Missions', color_activity, 'round')
create_arrow(ax, 5, 11.7, 5, 11)
create_box(ax, 5, 10.5, 1.8, 0.8, 'Complete\nMissions\n(Earn Money)', color_activity, 'round')
create_arrow(ax, 5, 10.1, 5, 9.9)

# RIGHT PATH - COLLECTION
create_box(ax, 7.5, 12.5, 1.8, 0.6, 'View Collection', color_activity, 'round')
create_arrow(ax, 7.5, 12.2, 7.5, 11.5)
create_box(ax, 7.5, 11, 1.8, 0.8, 'Organize/\nSell Cards', color_activity, 'round')
create_arrow(ax, 7.5, 10.6, 7.5, 9.9)

# 5. GAMEPLAY - Pack Opening (All paths converge)
create_box(ax, 5, 9.3, 2.5, 0.8, 'GAMEPLAY:\nOpen Pack\n(Mouse Control)', color_process, 'round')

# Arrows converging
create_arrow(ax, 2.5, 9.6, 4, 9.5)
create_arrow(ax, 7.5, 9.6, 6, 9.5)

# Arrow down
create_arrow(ax, 5, 8.9, 5, 8.2)

# 6. Pack Opening Animation
create_box(ax, 5, 7.8, 2.5, 0.6, 'Immersive Pack\nOpening Animation', color_activity, 'round')

# Arrow down
create_arrow(ax, 5, 7.5, 5, 6.8)

# 7. Reveal Cards
create_box(ax, 5, 6.4, 2.5, 0.6, 'Reveal Cards\n(Various Rarity)', color_activity, 'round')

# Arrow down
create_arrow(ax, 5, 6.1, 5, 5.4)

# 8. Player Control - Decision
create_box(ax, 5, 4.9, 2.5, 0.8, 'PLAYER CONTROL:\nKeep or Sell?', color_decision, 'diamond')

# Arrows
create_arrow(ax, 3.8, 4.6, 2.5, 4.2, 'Sell')
create_arrow(ax, 6.2, 4.6, 7.5, 4.2, 'Keep')

# Keep path
create_box(ax, 7.5, 3.8, 1.6, 0.6, 'Add to\nCollection', color_activity, 'round')
create_arrow(ax, 7.5, 3.5, 7.5, 3)
create_box(ax, 7.5, 2.6, 1.6, 0.6, 'Level Up\nCollection', color_activity, 'round')

# Sell path
create_box(ax, 2.5, 3.8, 1.6, 0.6, 'Sell Card', color_activity, 'round')
create_arrow(ax, 2.5, 3.5, 2.5, 3)
create_box(ax, 2.5, 2.6, 1.6, 0.6, 'Earn Money', color_activity, 'round')

# Arrows converging to decision
create_arrow(ax, 2.5, 2.3, 4, 1.8)
create_arrow(ax, 7.5, 2.3, 6, 1.8)

# 9. Check Completion
create_box(ax, 5, 1.4, 2.5, 0.8, 'Collection\nComplete?', color_decision, 'diamond')

# No - Continue playing
create_arrow(ax, 3.5, 1.4, 1.5, 1.4, 'No')
create_arrow(ax, 1.5, 1.4, 1.5, 14)
create_arrow(ax, 1.5, 14, 3.5, 14)

# Yes - Winning
create_arrow(ax, 5, 1, 5, 0.3, 'Yes')

# 10. GAME OVER - Winning
create_box(ax, 5, -0.2, 2.5, 0.7, 'GAME OVER:\nCollection Complete!\n(WINNING)', color_end, 'round')

# Add side note for losing condition
create_box(ax, 8.5, 6, 1.3, 1.2, 'Note:\nNo traditional\n"Losing".\nAlways can\nearn money\nfrom daily\nrewards', '#FFFACD', 'square')

plt.tight_layout()
plt.savefig('kaarten_flowchart.png', dpi=300, bbox_inches='tight', facecolor='white')
print("Flowchart saved as 'kaarten_flowchart.png'")
plt.show()
```
