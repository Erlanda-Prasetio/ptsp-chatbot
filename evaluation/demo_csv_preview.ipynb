# CSV Readability Test
Check how the scored results look in a Jupyter notebook

```python
import pandas as pd

# Load the demo scored results
df = pd.read_csv('demo_scored_results.csv')

print(f"📊 Total rows: {len(df)}")
print(f"📈 Accuracy: {df['is_correct'].mean()*100:.1f}%")
print()
```

```python
# Display key columns (what you'll analyze most)
key_cols = ['eval_id', 'dataset_source', 'category', 'query', 'answer_preview', 
            'is_correct', 'confidence_score', 'response_time_seconds', 'total_tokens']

df[key_cols]
```

```python
# Check answer preview length
df['answer_preview'].str.len().describe()
```

```python
# Summary by dataset source
df.groupby('dataset_source').agg({
    'is_correct': ['count', 'sum', 'mean'],
    'confidence_score': 'mean',
    'response_time_seconds': 'mean',
    'total_tokens': 'mean'
}).round(2)
```

```python
# Display settings for better readability
pd.set_option('display.max_colwidth', 80)  # Match our preview length
pd.set_option('display.width', 1000)

# View with better formatting
df[['eval_id', 'query', 'answer_preview', 'is_correct', 'confidence_score']]
```

## ✅ Readability Check

**Good indicators:**
- Answer preview fits in one line
- No excessive wrapping in notebook cells
- Key metrics visible without scrolling
- Easy to scan 50 rows
